A continuación se presenta el desglose agrupado de las columnas según su tipo de información:

**1. Identificación y Ubicación Geográfica**

**id_renabap:** Identificador único numérico asignado a cada barrio popular.

**nombre_barrio:** Nombre común con el que se identifica el barrio.

**provincia:** Provincia en la que se localiza el barrio.

**departamento:** Departamento, partido o comuna correspondiente.

**localidad:** Localidad o ciudad específica.

---------------------

**2. Demografía y Superficie**

**cantidad_viviendas_aproximadas:** Estimación del total de viviendas construidas en el barrio.

**cantidad_familias_aproximada:** Estimación de familias que residen en el barrio.

**superficie_m2:** Área total del barrio expresada en metros cuadrados ($m^2$).

---------------------

**3. Origen y Clasificación**

**decada_de_creacion:** Década aproximada en la que comenzó a formarse el asentamiento (ej. Década 1990, Década 2000).

**anio_de_creacion:** Año específico de creación (disponible para un subconjunto de barrios).

**clasificacion_barrio:** Tipo de asentamiento (ej. Asentamiento, Villa, Conjunto habitacional unifamiliar).

**titulo_propiedad:** Situación dominial que indica si el terreno posee título de propiedad regularizado (SI / NO).

---------------------

**4. Acceso a Servicios e Infraestructura**

**energia_electrica:** Tipo de conexión a la red eléctrica (conexión regular con medidor, irregular, etc.).

**efluentes_cloacales:** Sistema de eliminación de excretas y aguas servidas (cámara séptica, pozo negro/ciego, red cloacal).

**agua_corriente:** Modalidad de acceso al agua potable (conexión formal, irregular, bomba de pozo, etc.).

**cocina:** Principal recurso o combustible utilizado para cocinar (gas en garrafa, leña, red de gas, etc.).

**Calefaccion:** Principal recurso o combustible utilizado para calefaccionar la vivienda.

#**Capa Plata**


###1) Proteger el dato original

```python
# Creación de una copia explícita desde la capa bronce
df_renabap_plata = df_renabap.copy()
```

###2) Estandarizción de nombres

```python
# Renombrar 'nombre_barrio' a 'Barrio'
df_renabap_plata = df_renabap_plata.rename(
    columns={
        'nombre_barrio': 'barrio',
        'cantidad_familias_aproximada': 'cantidad_familias_aproximadas',
    }
)
```

```python
# Estandarizar nombres de columnas a snake_case
df_renabap_plata.columns = (
    df_renabap_plata.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('%', 'pct')
    .str.replace('á', 'a')
    .str.replace('é', 'e')
    .str.replace('í', 'i')
    .str.replace('ó', 'o')
    .str.replace('ú', 'u')
)
```

### 3) Normalización de Geografía y Cadenas Categóricas (Mayúsculas y sin tildes)

```python
cols_geograficas_renabap = ['provincia', 'departamento', 'localidad']
```

```python
for col in cols_geograficas_renabap:
  if col in df_renabap_plata.columns:
    df_renabap_plata[col] = (
        df_renabap_plata[col]
        .astype(str)
        .str.upper()
        .str.strip()
        .str.normalize('NFKD')
        .str.encode('ascii', errors='ignore')
        .str.decode('utf-8')
    )
print("Textos geográficos normalizados (Mayúsculas y sin tildes).")
```

```python
# Limpieza general del nombre del barrio (respetando formato lectura)
if 'barrio' in df_renabap_plata.columns:
  df_renabap_plata['barrio'] = (
      df_renabap_plata['barrio'].astype(str).str.strip()
  )
```

### 4) Tratamiento de Estructuras Especiales

En ReNaBaP no hay IDs multivaluados

### 5) Transformación de Tipos de Datos (Casting)

Tipo de dato de anio_de_creacion
La columna anio_de_creacion se importa como float64 (mostrando valores como 2018.0) debido a la presencia de NaN. Al exportar esto a Power BI o SQL, los años como flotantes pueden generar inconsistencias de formato.

Se recomienda castear la columna a un entero con soporte para nulos (Int64 de Pandas):

```python
df_renabap_plata['id_renabap'] = pd.to_numeric(
    df_renabap_plata['id_renabap'], errors='coerce'
).astype('Int64')

if 'anio_de_creacion' in df_renabap_plata.columns:
  df_renabap_plata['anio_de_creacion'] = pd.to_numeric(
      df_renabap_plata['anio_de_creacion'], errors='coerce'
  ).astype('Int64')
```

### 6) Estrategia de Imputación de Nulos y Categorías Especiales

```python
df_renabap_plata['barrio'] = df_renabap_plata['barrio'].fillna(
    'Sin Identificar'
)
```

```python
# B. Normalización de valores no relevados en servicios básicos
cols_servicios = [
    'agua_corriente',
    'energia_electrica',
    'efluentes_cloacales',
    'cocina',
    'calefaccion',
]

for col in cols_servicios:
  if col in df_renabap_plata.columns:
    # 1. Reemplazar cadenas 'Sin Datos' por 'SIN RELEVAMIENTO'
    df_renabap_plata[col] = df_renabap_plata[col].replace(
        'Sin Datos', 'SIN RELEVAMIENTO'
    )

    # 2. Reemplazar valores nulos (NaN) si existieran por la misma etiqueta en mayúsculas
    df_renabap_plata[col] = df_renabap_plata[col].fillna('SIN RELEVAMIENTO')

    # 3. Asegurar que todo el contenido de la columna quede en Mayúsculas y sin espacios residuales
    df_renabap_plata[col] = df_renabap_plata[col].astype(str).str.upper().str.strip()

print("Servicios básicos estandarizados con la etiqueta 'SIN RELEVAMIENTO'.")
```

###7) Enriquecimiento de Datos (Columnas Derivadas)

```python
# A. Promedio de familias por vivienda (evitando división por cero)
if (
    'cantidad_familias_aproximadas' in df_renabap_plata.columns
    and 'cantidad_viviendas_aproximadas' in df_renabap_plata.columns
):
  df_renabap_plata['familias_por_vivienda'] = (
      df_renabap_plata['cantidad_familias_aproximadas']
      / df_renabap_plata['cantidad_viviendas_aproximadas'].replace(0, np.nan)
  ).round(2)

# B. Densidad / Superficie por vivienda (m2_por_vivienda)
if (
    'superficie_m2' in df_renabap_plata.columns
    and 'cantidad_viviendas_aproximadas' in df_renabap_plata.columns
):
  df_renabap_plata['m2_por_vivienda'] = (
      df_renabap_plata['superficie_m2']
      / df_renabap_plata['cantidad_viviendas_aproximadas'].replace(0, np.nan)
  ).round(2)

print("Columnas derivadas calculadas exitosamente.")
```

###8) Observabilidad y Banderas de Calidad (Outlier Flags)

```python
# Identificar outliers por IQR sin borrar registros (Data Observability Flag)
if 'm2_por_vivienda' in df_renabap_plata.columns:
  q1 = df_renabap_plata['m2_por_vivienda'].quantile(0.25)
  q3 = df_renabap_plata['m2_por_vivienda'].quantile(0.75)
  iqr = q3 - q1

  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr

  # Crear columna booleana para filtrar fácilmente en Power BI
  df_renabap_plata['es_outlier_densidad'] = (
      df_renabap_plata['m2_por_vivienda'] < lower_bound
  ) | (df_renabap_plata['m2_por_vivienda'] > upper_bound)

  total_outliers = df_renabap_plata['es_outlier_densidad'].sum()
  print(
      f"Observabilidad: Se etiquetaron {total_outliers} registros como"
      " atípicos en densidad (es_outlier_densidad)."
  )
```

### 9) Deduplicación y Guardado en Parquet

```python
df_renabap_plata = df_renabap_plata.drop_duplicates(
    subset=['id_renabap'], keep='first'
).reset_index(drop=True)
```

```python
# Guardar en Parquet
df_renabap_plata.to_parquet(
    '/content/drive/MyDrive/Máster/TFM/renabap_plata.parquet', index=False
)
print(
    "Capa Plata RENABAP completada. Registros finales:",
    len(df_renabap_plata),
)
```


### 10) Obeservabilidad Final y Cruzabilidad con FISU


```python
from IPython.display import display

# Ruta al archivo Parquet de FISU (Capa Plata)
ruta_fisu_plata = '/content/drive/MyDrive/Máster/TFM/fisu_plata.parquet'

try:
    # 1. Cargar la capa Plata de FISU
    df_fisu_plata_obs = pd.read_parquet(ruta_fisu_plata)

    # 2. Extraer conjuntos de IDs únicos (ignorando nulos)
    # Usamos el df_renabap_plata que ya está en la memoria de este notebook
    ids_renabap = set(df_renabap_plata['id_renabap'].dropna())
    ids_fisu = set(df_fisu_plata_obs['id_renabap'].dropna())

    # 3. Cálculos de Cruzabilidad e Integridad mediante teoría de conjuntos
    ids_comunes = ids_fisu.intersection(ids_renabap)
    fisu_huerfanos = ids_fisu - ids_renabap  # IDs en FISU que no existen en el padrón ReNaBaP
    renabap_sin_fisu = ids_renabap - ids_fisu # Barrios ReNaBaP sin proyectos (Brecha de inversión)

    # Cálculo de Porcentajes
    match_rate_fisu = (len(ids_comunes) / len(ids_fisu)) * 100 if len(ids_fisu) > 0 else 0
    cobertura_renabap = (len(ids_comunes) / len(ids_renabap)) * 100 if len(ids_renabap) > 0 else 0

    # 4. Construir la Tabla de Métricas de Calidad
    df_observabilidad = pd.DataFrame({
        'Métrica de Calidad / Negocio': [
            'Total de Barrios únicos en ReNaBaP (Maestro)',
            'Total de IDs únicos en proyectos FISU (Transaccional)',
            'IDs de FISU validados en ReNaBaP (Match Exacto)',
            'IDs de FISU NO encontrados (Huérfanos / Error de Integridad)',
            'Barrios ReNaBaP SIN proyectos FISU (Brecha de Inversión)',
            '---',
            'Tasa de Integridad FISU (Match Rate)',
            'Tasa de Cobertura FISU sobre ReNaBaP'
        ],
        'Valor': [
            len(ids_renabap),
            len(ids_fisu),
            len(ids_comunes),
            len(fisu_huerfanos),
            len(renabap_sin_fisu),
            '---',
            f"{match_rate_fisu:.2f}%",
            f"{cobertura_renabap:.2f}%"
        ]
    })

    print("=========================================================")
    print("  REPORTE DE OBSERVABILIDAD DE DATOS (MATCH RATE)        ")
    print("=========================================================")
    display(df_observabilidad)

    # 5. Exportar el log de huérfanos para el Dashboard de Calidad en Power BI
    if len(fisu_huerfanos) > 0:
        df_errores = pd.DataFrame({'id_renabap_invalido': list(fisu_huerfanos)})
        df_errores.to_parquet('/content/drive/MyDrive/Máster/TFM/log_huerfanos_fisu.parquet', index=False)
        print(f"\n✅ Se exportaron {len(fisu_huerfanos)} IDs huérfanos a 'log_huerfanos_fisu.parquet' para visualización en Power BI.")
    else:
        print("\n✅ Excelente: No se detectaron IDs huérfanos. Integridad perfecta.")

except FileNotFoundError:
    print("⚠️ Aviso: No se encontró fisu_plata.parquet. Asegúrate de correr el notebook de FISU primero para poder calcular el Match Rate.")
```

#**Capa Oro**


Paso 1: Configuración inicial y carga de la Capa Plata

```python
ruta_renabap_plata = '/content/drive/MyDrive/Máster/TFM/renabap_plata.parquet'
ruta_fisu_plata = '/content/drive/MyDrive/Máster/TFM/fisu_plata.parquet'

df_renabap_plata = pd.read_parquet(ruta_renabap_plata)
df_fisu_plata = pd.read_parquet(ruta_fisu_plata)
```

Paso 2: Enriquecimiento Territorial (Creación de Regiones)
Definimos el diccionario de mapeo de provincias a regiones argentinas y la función estandarizada para clasificar cada registro.

```python
def asignar_region(provincia):
    provincia = str(provincia)
    if provincia in ['JUJUY', 'SALTA', 'TUCUMAN', 'CATAMARCA', 'LA RIOJA', 'SANTIAGO DEL ESTERO']:
        return 'NOA'
    elif provincia in ['FORMOSA', 'CHACO', 'CORRIENTES', 'MISIONES']:
        return 'NEA'
    elif provincia in ['CORDOBA', 'SANTA FE', 'ENTRE RIOS', 'BUENOS AIRES', 'CIUDAD AUTONOMA DE BUENOS AIRES', 'LA PAMPA']:
        return 'Centro'
    elif provincia in ['MENDOZA', 'SAN JUAN', 'SAN LUIS']:
        return 'Cuyo'
    elif provincia in ['NEUQUEN', 'RIO NEGRO', 'CHUBUT', 'SANTA CRUZ', 'TIERRA DEL FUEGO']:
        return 'Patagonia'
    else:
        return 'Sin Asignar'

if 'provincia' in df_renabap_plata.columns:
    df_renabap_plata['Region'] = df_renabap_plata['provincia'].apply(asignar_region)
```

Paso 3. Selección de clumnas de la dimensión: incluimos las banderas de calidad y las métricas calculadas en Plata.

```python
columnas_dimension = [
    'id_renabap',
    'barrio',
    'Region',
    'provincia',
    'departamento',
    'localidad',
    'cantidad_familias_aproximadas',
    'familias_por_vivienda',
    'm2_por_vivienda',
    'es_outlier_densidad'
]

# Filtramos solo las columnas que existan para evitar errores
cols_presentes = [col for col in columnas_dimension if col in df_renabap_plata.columns]
df_renabap_oro = df_renabap_plata[cols_presentes].copy()
```

Paso 4: Inyección de huérfanos FISU y registro desconocido.
Integramos de forma limpia el proceso para detectar y rescatar aquellos barrios que existen en FISU pero faltaban en ReNaBaP, evitando inconsistencias analíticas en Power BI.

```python
ids_fisu = set(df_fisu_plata['id_renabap'].dropna())
ids_renabap = set(df_renabap_oro['id_renabap'].dropna())
huerfanos = ids_fisu - ids_renabap

filas_adicionales = []

# A. Crear registros para los IDs de FISU que no existen en ReNaBaP
for id_huerfano in huerfanos:
    filas_adicionales.append({
        'id_renabap': id_huerfano,
        'barrio': 'Desconocido (Huérfano FISU)',
        'Region': 'Sin Asignar',
        'provincia': 'SIN PROVINCIA',
        'departamento': 'SIN DEPARTAMENTO',
        'localidad': 'SIN LOCALIDAD',
        'cantidad_familias_aproximadas': 0,
        'familias_por_vivienda': 0.0,
        'm2_por_vivienda': 0.0,
        'es_outlier_densidad': False
    })

# B. Crear el Registro Comodín (99999) para proyectos FISU sin ID asignado
filas_adicionales.append({
    'id_renabap': 99999,
    'barrio': 'Sin Asignar / Desconocido',
    'Region': 'Sin Asignar',
    'provincia': 'SIN PROVINCIA',
    'departamento': 'SIN DEPARTAMENTO',
    'localidad': 'SIN LOCALIDAD',
    'cantidad_familias_aproximadas': 0,
    'familias_por_vivienda': 0.0,
    'm2_por_vivienda': 0.0,
    'es_outlier_densidad': False
})

# Concatenar las filas adicionales al dataset principal
if filas_adicionales:
    df_adicionales = pd.DataFrame(filas_adicionales)
    # Asegurarnos de que el orden y las columnas coincidan
    df_adicionales = df_adicionales[cols_presentes]
    df_renabap_oro = pd.concat([df_renabap_oro, df_adicionales], ignore_index=True)
```

Paso 5: Presentación semántica: Nombres para Power Bi

```python
diccionario_nombres = {
    'id_renabap': 'ID ReNaBaP',
    'barrio': 'Nombre del Barrio',
    'Region': 'Región Geográfica',
    'provincia': 'Provincia',
    'departamento': 'Departamento',
    'localidad': 'Localidad',
    'cantidad_familias_aproximadas': 'Cantidad de Familias',
    'familias_por_vivienda': 'Familias por Vivienda',
    'm2_por_vivienda': 'm² por Vivienda (Densidad)',
    'es_outlier_densidad': 'Es Atípico en Densidad'
    }

df_renabap_oro = df_renabap_oro.rename(columns=diccionario_nombres)
```
